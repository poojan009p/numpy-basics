## Advance array Manipulations 

---


#### Stacking array

```python
b3 = np.arange(1,6)
b4 = np.arange(6,11)  #Here we cannot use matrix we have to use vector form only.
b3 ,b4
```

```python
# now lets stack it vertically horizontally and normal stack
```

```python
np.vstack((b3,b4))
```

```python
np.hstack((b3,b4))
```

```python
np.column_stack((b3,b4))
```
