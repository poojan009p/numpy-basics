### Indexing and Slicing of Vectors 

---

```python
a1 = np.arange(1,10)
a1
```

```python
a1[8]  #box bracket is used to call index value 
```

```python
a1[0:5]
```

```python
a1[5:]
```

```python
aslice = a1[3:6]
aslice
```

### Indexing and Slicing of Matrix 

---

```python
a2 = np.arange(1,31).reshape(6,5)
a2
```

```python
a2[3]             # It gives value on x axis like in cartesian plane 
```

```python
a2[0:6,0:2]       # It gives values on y axis by slicing.
```

```python
a2[2:4,2:4]        # Small Sliced part of full matrix. 
```

```python
a2[3:5,1:4]        # 2*3 matrix sliced from main 6*5 matrix and we can also save this sliced value in different variable
```
